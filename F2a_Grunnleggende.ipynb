## Datatyper

En datatype beskriver hva slags verdi noe er.
- `int` = heltall
- `float` = desimaltall
- `str` = tekst
- `list` = lister
- `dict` = ordbøker


```python
x = 42
y = 3.14
s = "Hei"
liste = [1, 2, 3]
d = {"navn": "Ola", "alder": 25}

print(type(x), type(y), type(s), type(liste), type(d))
```

## Objekter

Et objekt er en konkret instans av en datatype eller klasse.
- `42` er et objekt av typen `int`.
- `[1,2,3]` er et objekt av typen `list`.

Alt i Python er objekter!

## Klasser

En klasse er en "oppskrift" for objekter. Den beskriver hvilke data og metoder objektene har.

Eksempel: vi lager en enkel klasse `Hund`.

```python
class Hund:
    def __init__(self, navn, alder):
        self.navn = navn
        self.alder = alder
    
    def bjeff(self):
        print(f"{self.navn} sier: Voff!")

h1 = Hund("Fido", 3)
h1.bjeff()
print(type(h1))
```

## NumPy Arrays

NumPy gir oss en effektiv datatype: `ndarray`.
Dette er en N-dimensjonal array som er mye raskere og mer fleksibel enn vanlige Python-lister for numeriske beregninger.

```python
import numpy as np

a = np.array([1, 2, 3])
print(a)
print(type(a))

#En vanlig liste i Python: 
#b = [1, 2, 3, 4]
#print(b)
#print(type(b))
# <class 'list'>
```

## Printing i Python

- Bruk `print("tekst")` for å skrive ut en streng.
- For å kombinere tekst og tall:
  - bruk `,` i `print`
  - eller f-strenger (`f"Alder: {alder}")`

```python
alder = 25
print("Alder:", alder)
print(f"Alder: {alder}")
```
