```python
# --- Imports og feltdefinisjon ---
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401 (nødvendig for 3D)

def F(x, y, z, z_eps=1e-6):
    """
    Vektorfelt: F = (-y/z, x/z, 0). Returnerer komponenter som numpy-arrays.
    Unngår deling på ~0 ved å maskere punkter der |z| < z_eps.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    z = np.asarray(z, dtype=float)

    Fx = np.zeros_like(x, dtype=float)
    Fy = np.zeros_like(y, dtype=float)
    Fz = np.zeros_like(z, dtype=float)

    mask = np.abs(z) >= z_eps
    Fx[mask] = -y[mask] / z[mask]
    Fy[mask] =  x[mask] / z[mask]
    Fz[...]  = 0.0
    # Sett NaN der feltet er udefinert (nær z=0), slik at de ikke tegnes
    Fx[~mask] = np.nan
    Fy[~mask] = np.nan
    return Fx, Fy, Fz

# --- 1) 3D-plott av feltet ---
# Velg et rutenett som ikke går gjennom z=0
x = np.linspace(-2, 2, 9)
y = np.linspace(-2, 2, 9)
z = np.linspace(0.5, 2.0, 6)   # kun positive z (du kan lage et eget sett for negative z også)

X, Y, Z = np.meshgrid(x, y, z, indexing="xy")
U, V, W = F(X, Y, Z)

# For oversikt: tynn ut piler (valgfritt)
step = (slice(None, None, 2), slice(None, None, 2), slice(None, None, 2))
Xq, Yq, Zq = X[step], Y[step], Z[step]
Uq, Vq, Wq = U[step], V[step], W[step]

fig = plt.figure(figsize=(7,6))
ax = fig.add_subplot(111, projection='3d')
ax.quiver(Xq, Yq, Zq, Uq, Vq, Wq, length=0.3, normalize=True)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title(r'3D vektorfelt $\,\mathbf F=(-y/z,\;x/z,\;0)$ (maskert nær $z=0$)')
ax.set_box_aspect((1,1,0.7))
plt.show()

# --- 2) 2D-snitt ved z = 1 ---
z0 = 1.0
x2 = np.linspace(-3, 3, 25)
y2 = np.linspace(-3, 3, 25)
X2, Y2 = np.meshgrid(x2, y2, indexing="xy")
U2, V2, _ = F(X2, Y2, z0*np.ones_like(X2))

fig, ax = plt.subplots(figsize=(6,6))
ax.quiver(X2, Y2, U2, V2, alpha=0.8, pivot='mid')
# (valgfritt) legg til strømlinjer for å se sirkulasjonen tydelig
ax.streamplot(x2, y2, U2, V2, density=1.2, linewidth=0.8, arrowsize=1.2)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_aspect('equal', adjustable='box')
ax.set_title(r'Snitt $z=1$: $\mathbf F(x,y,1)=(-y,\;x,\;0)$')
ax.grid(True, alpha=0.3)
plt.show()
```
