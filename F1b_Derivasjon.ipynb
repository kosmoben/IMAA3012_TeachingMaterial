# Repetisjon: Partiell derivasjon

Vi kan bruke sympy for å regne ut

```python
import sympy as sp

def deriver_xyz():
    # Definer symbolene
    x, y, z = sp.symbols('x y z')
    
    # Funksjonen
    f = x * y * z
    
    # Partiellderivasjoner
    df_dx = sp.diff(f, x)
    df_dy = sp.diff(f, y)
    df_dz = sp.diff(f, z)
    
    return df_dx, df_dy, df_dz

# Kjør funksjonen
Df_Dx, Df_Dy, Df_Dz = deriver_xyz()
print("∂f/∂x =", Df_Dx)
print("∂f/∂y =", Df_Dy)
print("∂f/∂z =", Df_Dz)
```

```python
import numpy as np

def numerisk_partiellderiv(f, x0, y0, z0, h=1e-5):
    """
    Beregner numeriske partiellderivasjon av f(x,y,z) ved punkt (x0, y0, z0)
    h er liten steglengde for finitt differanse
    """
    # ∂f/∂x
    dfdx = (f(x0 + h, y0, z0) - f(x0 - h, y0, z0)) / (2*h)
    
    # ∂f/∂y
    dfdy = (f(x0, y0 + h, z0) - f(x0, y0 - h, z0)) / (2*h)
    
    # ∂f/∂z
    dfdz = (f(x0, y0, z0 + h) - f(x0, y0, z0 - h)) / (2*h)
    
    return dfdx, dfdy, dfdz

# Definer funksjonen
f = lambda x, y, z: x*y*z

# Eksempel: punkt (1,2,3)
dfdx, dfdy, dfdz = numerisk_partiellderiv(f, 1, 2, 3)
print("∂f/∂x =", dfdx)
print("∂f/∂y =", dfdy)
print("∂f/∂z =", dfdz)
```
