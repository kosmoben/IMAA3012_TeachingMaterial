La oss si at funksjonen vår er $f(x)=\cos(x)$. Vi ønsker å regne ut $$\displaystyle{\int_R f(x)\; dx}$$ for 
$$ -1 \leq x < 1 .$$

## Numerisk bibliotek (Numpy = numerisk Python)

Vi gir en enkel numerisk metode hvor vi legger et gitter over integrasjonsområdet, regner ut funksjonsverdier og summerer. 

```python
import numpy as np

# Arealelementene
dx = 0.000001

# Gitterpunkter
x = np.arange(-np.pi, np.pi, dx)

# Integrasjon
result = np.sum(np.cos(x)*dx)

print(result)
```

## Symbolsk utregning (Sympy= symbolic python)

```python
import sympy as sp

def integrasjon_sympy():
    x = sp.symbols('x')
    f = sp.cos(x)
    
    # Symbolsk integral
    integral_symbolsk = sp.integrate(f, (x, -sp.pi, sp.pi))
    
    # Numerisk verdi (evaluerer sympy-uttrykket)
    integral_numerisk = float(integral_symbolsk.evalf())
    
    return integral_symbolsk, integral_numerisk

symbolsk, numerisk = integrasjon_sympy()
print("Symbolsk resultat:", symbolsk)
print("Numerisk verdi:", numerisk)
```

## Scipy (Scientific Python)

```python
from scipy import integrate
import numpy as np

f = lambda x: np.sin(x)
(eval, error) = integrate.quad(f, 0, np.pi) #Her er (eval,error) en tupple i Python ( en datatype).

print(eval)   # ≈ 2.0
print(error)    # estimerte feil
```
