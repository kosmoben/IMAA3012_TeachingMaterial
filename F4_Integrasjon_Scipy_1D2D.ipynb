## Scipy for 2D

Utregning av integralet $$\displaystyle{\int_c^d\int_{a(x)}^{b(x)} f(x,y)\; dx\; dy}$$ kan også gjøres med Scipy sin funksjon quad. Funksjonen vi da bruker heter dblquad og er altså en høynivå funksjon. Den
* Benytter algoritmen QUADPACK, som er en adaptiv Gauss-Kronod kvadraturmetode (deler opp i små viter og evalurer funksjonen et sted på hver bit. 
* Stopper evalueringen når feilestimatet er under en gitt toleranseverdi
* `dblquad(f, a, b, gfun, hfun)` betyr:$$
\int_a^b \left( \int_{g(x)}^{h(x)} f(x,y)\, dy \right) dx
$$

Her er $$a = -1, \, b = 1$$, og $$gfun = -1, \, hfun = 1$$.


## Eksempel
La oss som eksempel regne ut integralet 

$$\int_0^1 \left( \int_x^{\sqrt{x}} x+y\, dy \right) dx$$

ved hjelp av Scipy:


```python
import numpy as np
from scipy import integrate

f = lambda y, x : x*y
a = lambda x : x
b = lambda x : np.sqrt(x)
c = 0
d = 1

result = integrate.dblquad(f, c, d, a, b)

print(result)
print(type(result))
print(result[0], "er her integralet evaluert, mens") # value of the integral
print(result[1], "er feilestimatet") # estimate of error

```

# Et lite triks...

Det kan også være verdt å merke seg et lite numerisk triks vi kan benytte når vi bruker et dataverktøy som Python. Vi kan nemlig regne ved å integrere over større områder og forkaste verdier som ikke
skal brukes.

Her regner vi ut integralet $$\displaystyle{\iint_D x + y\; dA}$$ for disken $$D=\{(x,y)\in \mathbb{R}^2\; | \; x^2+y^2\leq 1\}.$$

Vi gjør dette ved å integrere over et rektangel som inneholder $D$, 
$$\displaystyle{\int_{-1}^1\int_{-1}^1 f(x,y)\; dA}$$
men definerer funksjonen $f$ lik $0$ utenfor disken. 


```python
import numpy as np
from scipy import integrate

f = lambda x,y : x + y if x*x+y*y <= 1 else 0

result = integrate.dblquad(f, -1,1, -1,1)
print(result)
```
