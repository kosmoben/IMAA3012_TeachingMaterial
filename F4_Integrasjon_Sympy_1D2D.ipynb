# Sympy for 1D

## Eksempel 1 [Ø2, O10]

```python
import sympy as sp
x,y = sp.symbols("x,y")

f = 2*x*x + y

svar = sp.integrate(f,(x,0,1),(y,-2,2))
print(svar)
```

## Eksempel 2
La oss si at funksjonen vår er $f(x)=\cos(x)$. Vi ønsker å regne ut $$\displaystyle{\int_R f(x)\; dx}$$ for 
$$ -1 \leq x < 1 .$$

```python
import sympy as sp

def integrasjon_sympy():
    x = sp.symbols('x')
    f = sp.cos(x)
    
    # Symbolsk integral
    integral_symbolsk = sp.integrate(f, (x, -sp.pi, sp.pi))
    
    # Numerisk verdi (evaluerer sympy-uttrykket)
    integral_numerisk = float(integral_symbolsk.evalf())
    
    return integral_symbolsk, integral_numerisk

symbolsk, numerisk = integrasjon_sympy()
print("Symbolsk resultat:", symbolsk)
print("Numerisk verdi:", numerisk)
```

# Sympy for 2D

Sympy kan også regne ut dobbeltintegraler direkte -- vi gir da bare to `(variabel, nedre, øvre)`-par til `sp.integrate`:
$$\displaystyle{\int_c^d\int_a^b f(x,y)\; dx\, dy}$$

## Eksempel

La oss som eksempel regne ut integralet

$$\int_0^1\int_0^1 (x^2+y^2)\; dx\, dy$$

ved hjelp av Sympy:

```python
import sympy as sp

x, y = sp.symbols('x y')
f = x**2 + y**2

svar = sp.integrate(f, (x, 0, 1), (y, 0, 1))
print(svar)
```

### Variable grenser på det innerste integralet

Så langt har vi integrert over rektangler, med konstante grenser. Ofte er området vi integrerer over ikke et rektangel -- da vil grensen på det innerste integralet avhenge av den andre variabelen. I Sympy er ikke dette noe vanskeligere: vi skriver bare grensen som et uttrykk i $x$ i stedet for et tall.

#### Eksempel

La oss regne ut arealet av trekanten mellom $y=0$ og $y=x$, for $0\le x\le 1$ (altså $f(x,y)=1$):
$$\int_0^1\int_0^x 1\, dy\, dx.$$

```python
import sympy as sp

x, y = sp.symbols('x y')
f = 1

svar = sp.integrate(f, (y, 0, x), (x, 0, 1))   # merk: øvre grense for y er "x", ikke et tall
print(svar)
```
